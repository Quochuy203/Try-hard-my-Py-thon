Bài toán Viết hàm trả về một dictionary đếm số lượng chữ xuất hiện
trong một từ, với key là chữ cái và value là số lần xuất hiện
• Input: một từ
• Output: dictionary đếm số lần các chữ xuất hiện
• Note: Giả sử các từ nhập vào đều có các chữ cái thuộc [a-z] hoặc [A-Z]

Résoudre l'exo avec niveau de junior

```python
import time

def count_character(word):
    result = {}
    for character in word:
        if character in result:
            result[character] += 1
        else:
            result[character] = 1
    return dict(sorted(result.items()))

start = time.perf_counter()
end = time.perf_counter()


print(count_character("Happiness"))
print(f"Run in {(end - start) * 1_000_000:.4f}s")
```

Résoudre l'exo avec niveau de mid-lv
- Utiliser pas "if", essayer de utiliser .get()


```python
import time

def count_word(word: str) -> dict[str: int]:
    ## Comptez le nombre de fois où chaque lettre apparaît (sensible à la casse).
    if not isinstance(word, str):
        raise TypeError(f"word isn't string, recu {type(word).__name__}")
    if word and not (word.isalpha() and word.isascii):
        raise ValueError("Juste acceptée de a-z et A-Z")
    count = {}
    for ch in word:
        count[ch] = count.get(ch, 0) + 1
    return dict(sorted(count.items()))

start = time.perf_counter()
end = time.perf_counter()

print(count_word("Happyness"))
print(f"Time Run in {(end - start) * 1_000_000:.4f}")

### Try with false value
try:
    count_word("Il fait beau")
except ValueError as e:
    print("Trouvé l'erreur:", e)
```

```python
def count_letters(text: str) -> dict[str, int]:
    if not isinstance(text, str):
        raise TypeError(f"text phải là str, nhận được {type(text).__name__}")
    count = {}
    for ch in text:
        if ch.isascii() and ch.isalpha():   # kiểm tra TỪNG ký tự, bỏ qua ký tự không hợp lệ
            count[ch] = count.get(ch, 0) + 1
    return dict(sorted(count.items()))

print(count_letters("Il fait beau"))
```
