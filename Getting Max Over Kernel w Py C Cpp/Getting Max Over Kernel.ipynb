Trouver la plus grande valeur dans la fenêtre glissante d'une liste donnée

- Problème : Étant donné une liste d'entiers num_list et une fenêtre glissante de taille k se déplaçant de gauche à droite. À chaque déplacement de la fenêtre d'une position vers la droite, k nombres de num_list sont visibles. Trouver le plus grand nombre parmi ces k nombres, après chaque glissement, supérieur ou égal à 1.
- input: num_list = [3, 4, 5, 1, -44 , 5 ,10, 12 ,33, 1] avec  k=3
- Output: [5, 5, 5, 5, 10, 12, 33, 33]
- Exemple: 
[3, 4, 5], 1, -44 , 5 ,10, 12 ,33, 1 => max 5
3, [4, 5, 1], -44 , 5 ,10, 12 ,33, 1 => max 5
3, 4, [5, 1, -44] , 5 ,10, 12 ,33, 1 => max 5
3, 4, 5, [1, -44 , 5] ,10, 12 ,33, 1 => max 5
3, 4, 5, 1, [-44 , 5 ,10], 12 ,33, 1 => max 10
3, 4, 5, 1, -44 , [5 ,10, 12] ,33, 1 => max 12
3, 4, 5, 1, -44 , 5 ,[10, 12 ,33], 1 => max 33
3, 4, 5, 1, -44 , 5 ,10, [12 ,33, 1] => max 33

```python
### Travaille cet exo avec le niveau Junior
import time

def max_kernel(num_list, k):
    result = []
    for i in range (len(num_list)-k+1):
        result.append(max(num_list[i:i+k]))
    return result

num_list = [3, 4, 5, 1, -44 , 5 ,10, 12 ,33, 1]

start = time.perf_counter()
end = time.perf_counter()

print(max_kernel(num_list, 3))
print(f"Run in: {(end-start)*1_000_000:.4f} µs")
```

```python
### Travaille cet exo avec le niveau +1
def max_kernel(num_list, k):
    result = []
    print("len =", len(num_list), "→ số vòng lặp =", len(num_list) - k + 1)
    for i in range(len(num_list) - k + 1):
        window = num_list[i:i + k]
        biggest = max(window)
        result.append(biggest)
        print(f"i={i} | num_list[{i}:{i+k}] = {window} | max = {biggest} | result = {result}")
    return result

num_list = [3, 4, 5, 1, -44, 5, 10, 12, 33, 1]
print(max_kernel(num_list, 3))
```

```python
### Mid-level
from typing import List

def max_kernel(num_list: List[int], k: int) -> List[int]:
    if not isinstance(k, int) or k <= 0:
        raise ValueError(f"k is not so nguyen duong, nhan duoc {k!r}")
    if k > len(num_list):
        return[]
    return [max(num_list[i:i + k]) for i in range(len(num_list) - k + 1)]

num_list = [3, 4, 5, 1, -44, 5, 10, 12, 33, 1]
print(max_kernel(num_list, 3))
```
