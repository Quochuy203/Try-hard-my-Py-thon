Problème : Écrire une fonction qui lit des phrases depuis un fichier texte, compte le nombre d'occurrences de chaque mot et renvoie un dictionnaire dont la clé est le mot et la valeur son nombre d'occurrences.

• Entrée : Chemin d'accès au fichier texte

• Sortie : Dictionnaire contenant le nombre d'occurrences de chaque mot

• Remarque :

– On suppose que tous les mots du fichier texte sont des lettres de l'alphabet [a-z] ou [A-Z]

– Aucune manipulation complexe de chaînes de caractères n'est nécessaire, mais il faut tenir compte du fait que tous les mots sont en minuscules.
