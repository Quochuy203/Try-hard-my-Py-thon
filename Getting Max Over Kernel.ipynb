Trouver la plus grande valeur dans la fenêtre glissante d'une liste donnée

- Problème : Étant donné une liste d'entiers num_list et une fenêtre glissante de taille k se déplaçant de gauche à droite. À chaque déplacement de la fenêtre d'une position vers la droite, k nombres de num_list sont visibles. Trouver le plus grand nombre parmi ces k nombres, après chaque glissement, supérieur ou égal à 1.
- input: num_list = [3, 4, 5, 1, -44 , 5 ,10, 12 ,33, 1] avec  k=3
- Output: [5, 5, 5, 5, 10, 12, 33, 33]
- Exemple: 
[3, 4, 5], 1, -44 , 5 ,10, 12 ,33, 1 => max 5
3, [4, 5, 1], -44 , 5 ,10, 12 ,33, 1 => max 5
3, 4, [5, 1, -44] , 5 ,10, 12 ,33, 1 => max 5
3, 4, 5, [1, -44 , 5] ,10, 12 ,33, 1 => max 5
3, 4, 5, 1, [-44 , 5 ,10], 12 ,33, 1 => max 10
3, 4, 5, 1, -44 , [5 ,10, 12] ,33, 1 => max 12
3, 4, 5, 1, -44 , 5 ,[10, 12 ,33], 1 => max 33
3, 4, 5, 1, -44 , 5 ,10, [12 ,33, 1] => max 33

### Travaille cet exo avec le niveau Junior
def max_kernel(num_list, k):
    result = []
    for i in range (len(num_list)-k):
        result.append(max(num_list[i:i+k]))
    return result
